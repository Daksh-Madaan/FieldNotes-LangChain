### Generative AI Application using Agents and Google GenAI

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage, AIMessage
from langchain.tools import tool

@tool("HashFunction", description="Returns a hashed value of a given string, uses a custom hash function")
def example(arg: str) -> str:
    key = 0
    for i in arg:
        key = key*31 + ord(i)
    return str(key % 1000000007)

agent = create_agent(model="google_genai:gemini-3.6-flash", 
            tools=[example],
            system_prompt="When asked for a hashed string, use the HashFunction tool")


messages = [
    HumanMessage("Give me the hash version of 'hello'"),
    AIMessage("Here's Your Hashed String: 99162322"),
    HumanMessage("Give me the hash version of 'Daksh Madaan'")
]

response = agent.invoke({"messages": messages})

for m in response["messages"]:
    m.pretty_print()
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(model="google_genai:gemini-3.6-flash")
result = model.invoke("Write about how Sam Altman is actually an Alien")

print(result.content)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

m = ChatGoogleGenerativeAI(model = "gemini-3.6-flash")
result2 = m.invoke("How many times has Sam Altman lied on record?")
print(result2.content[0]['text'])
```


### Streaming and Batch

```python
result2 = m.stream("Fetch me some proofs that OpenAI is actually funded by the KKK")
print(result2)

for chunk in result2:
    print(chunk.text,end="|",flush=True)
```

Returns a generator object (similar to an Iterator), and should be used with a Loop

```python
result2 = m.batch([
    "If Github were a Pokemon, What Pokemon would it be?",
    "How cooked is the Canada if they get invaded by penguins?",
    "Has anyone tried to rename Europe?"
    ],
    config={
        'max_concurrency': 2,
    }
)

for resp in result2:
    print(resp)
```
