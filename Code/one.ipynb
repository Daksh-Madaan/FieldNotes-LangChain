# Setting up keys for LangChain 

```python
import os
from dotenv import load_dotenv

load_dotenv()

os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

### Generative AI Application using Agents and Google GenAI

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage, AIMessage
from langchain.tools import tool

@tool("HashFunction", description="Returns a hashed value of a given string, uses a custom hash function")
def example(arg: str) -> str:
    key = 0
    for i in arg:
        key = key*31 + ord(i)
    return str(key % 1000000007)

agent = create_agent(model="google_genai:gemini-3.6-flash", 
            tools=[example],
            system_prompt="When asked for a hashed string, use the HashFunction tool")


messages = [
    HumanMessage("Give me the hash version of 'hello'"),
    AIMessage("Here's Your Hashed String: 99162322"),
    HumanMessage("Give me the hash version of 'Daksh Madaan'")
]

response = agent.invoke({"messages": messages})

for m in response["messages"]:
    m.pretty_print()
```
